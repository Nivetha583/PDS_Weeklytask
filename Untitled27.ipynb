```python
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python
df["total_stay"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

print(df[["stays_in_weekend_nights",
          "stays_in_week_nights",
          "total_stay"]].head())
```

```python
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

print(df[["adults", "children",
          "babies", "total_guests"]].head())
```

```python
# Cell 22
cancellation_rate = df["is_canceled"].mean() * 100

print("Cancellation Rate:", cancellation_rate, "%")

# Cell 23
hotel_cancel = df.groupby("hotel")["is_canceled"].mean() * 100

print(hotel_cancel)
```

```python
for col in ["lead_time", "adr"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]
    
    print(col, "Outliers:", len(outliers))
```

```python
sns.countplot(data=df, x="hotel")

plt.title("City Hotel vs Resort Hotel Bookings")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")

plt.show()
```

```python
sns.barplot(
    x=hotel_cancel.index,
    y=hotel_cancel.values
)

plt.title("Cancellation Rate by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Cancellation Rate (%)")

plt.show()
```

```python
sns.boxplot(
    data=df,
    x="is_canceled",
    y="lead_time"
)

plt.title("Lead Time vs Cancellation")
plt.xlabel("Cancellation (0 = No, 1 = Yes)")
plt.ylabel("Lead Time")

plt.show()
```

```python
print(df.columns.tolist())
```

```python
monthly_adr = df.groupby(
    "arrival_date_month"
)["adr"].mean()

months = [
    "January", "February", "March",
    "April", "May", "June",
    "July", "August", "September",
    "October", "November", "December"
]

monthly_adr = monthly_adr.reindex(months)

plt.figure(figsize=(12, 5))

plt.plot(months, monthly_adr.values, marker="o")

plt.title("Monthly Average Daily Rate")
plt.xlabel("Month")
plt.ylabel("Average Daily Rate")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="is_canceled",
    y="lead_time"
)

plt.title("Lead Time Distribution")
plt.xlabel("Cancellation Status")
plt.ylabel("Lead Time")

plt.show()
```

```python
segment_cancel = df.groupby(
    "market_segment"
)["is_canceled"].mean() * 100

segment_cancel.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Cancellation Rate by Market Segment")
plt.xlabel("Market Segment")
plt.ylabel("Cancellation Rate (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Cell [43]
plt.figure(figsize=(10, 5))

plt.hist(df["lead_time"], bins=20, edgecolor="black")

plt.title("Distribution of Lead Time")
plt.xlabel("Lead Time")
plt.ylabel("Frequency")

plt.show()
```

```python
# Cell [40]
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Hotel Booking Correlation Heatmap")

plt.show()
```
